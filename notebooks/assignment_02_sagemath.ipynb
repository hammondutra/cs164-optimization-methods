**SageMath notebook.** Models the cross-sectional area of a three-meter metal-sheet design, plots the feasible region, computes symbolic partial derivatives, and locates the interior stationary point. The full mathematical argument, Hessian test, and boundary comparison are in the [final report](../assignments/assignment_02_final_report.pdf).

The objective is $A(a,\theta)=a(3-a-a\sin\theta)\sin\theta+\frac12a^2\sin\theta\cos\theta$, with $0<a<\frac{3}{1+\sin\theta}$ and $0\leq\theta\leq\pi/2$.

**Environment:** Run in a SageMath Jupyter kernel (not a standard Python kernel). Execute cells from top to bottom.

## 1. Surface plot

Parameterize the feasibility boundary to keep the full surface within the metal-length constraint.

```sage
# Code for the surface plot:
var('a theta u')

f(a, theta) = a*(3 - a - a*sin(theta))*sin(theta) + a^2*sin(theta)*cos(theta)/2

A = 3*u/(1 + sin(theta))

parametric_plot3d(
    (A, theta, f(A, theta)),
    (u, 0, 1),
    (theta, 0, pi/2),
    plot_points=[60, 60],
    axes_labels=['a', 'theta', 'Area'])
```

## 2. Contour plot

Plot the objective over the feasible $(a,\theta)$ region.

```sage
# Code for contour_plot():

contour_plot(
    f(a, theta),
    (a, 0, 3),
    (theta, 0, pi/2),
    region=3 - a*(1 + sin(theta)),
    plot_points=300,
    contours=30,
    cmap='viridis',
    colorbar=True,
    axes_labels=['a (m)', 'theta (radians)'])
```

## 3. Symbolic differentiation

Compute $\partial A/\partial a$ and $\partial A/\partial\theta$.

```sage
# Code for computing partial derivatives of A(a, θ)):

# Partial derivative with respect to a
df_da = diff(f(a, theta), a)

# Partial derivative with respect to theta
df_dtheta = diff(f(a, theta), theta)

# Display simplified derivatives
show(df_da.simplify_full())
show(df_dtheta.simplify_full())
```

## 4. Numerical stationary-point solve

Reduce the two gradient equations to one equation in $\theta$ and solve with SageMath's `find_root`.

```sage
# Code to solve for value where ∂A/∂a = 0, ∂A/∂theta = 0

var('a theta u')

f(a, theta) = a*(3 - a - a*sin(theta))*sin(theta) + a^2*sin(theta)*cos(theta)/2

A = 3*u/(1 + sin(theta))

df_da = diff(f(a, theta), a)
df_dtheta = diff(f(a, theta), theta)

# Express a in terms of theta
a_expr = 3/(2 + 2*sin(theta) - cos(theta))

# Substitute into the second partial derivative
equation = df_dtheta.subs(a=a_expr)

# Find theta numerically
theta_sol = find_root(equation, 0.1, pi/2 - 0.1)

# Recover a
a_sol = a_expr.subs(theta=theta_sol).n()

show(theta_sol)
show(a_sol)
```
