````python
def f(x):
    return (2*x + 1)**2

x = np.linspace(-2, 1, 400)
plt.figure(figsize=(8, 4))
plt.plot(x, f(x), label=r'$f(x)=(2x+1)^2$')
plt.xlabel('x'); plt.ylabel('f(x)')
plt.legend(); plt.grid(True); plt.show()
````

````python
def g(x, y):
    return 2*x - 3*y + y**2

x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, y)
Z = g(X, Y)
plt.figure(figsize=(6, 6))
contours = plt.contourf(X, Y, Z, levels=20, cmap='Blues')
plt.colorbar(contours, label='g(x,y)')
plt.xlabel('x'); plt.ylabel('y'); plt.title('Contour map')
plt.show()
````

**Interpretation:** the first plot shows a parabola with its minimum at $x=-1/2$. The second illustrates contour plotting, which will be reused later for quadratic forms and descent directions.
