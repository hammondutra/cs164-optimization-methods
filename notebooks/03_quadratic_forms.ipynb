## Question 2 — Calculus review

## Question 3 — Why is linear regression an unconstrained optimization problem?
**Submitted response:** choose linear-model parameters to minimize squared error over a dataset. A simple linear model can be written $\hat y_i=\alpha x_i+\beta$; its objective is $\min_{\alpha,\beta}\sum_i(y_i-\hat y_i)^2$ when the parameters are unrestricted.


## Question 4 — Necessary versus sufficient
**Submitted response:** a necessary condition must hold for the target outcome, but may not imply that outcome; a sufficient condition guarantees it. Thus $f'(c)=0$ is necessary for an unconstrained interior local optimum of a differentiable function but is not sufficient (it could be a maximum or saddle).


## Question 5 — Quadratic approximation examples
The visible submission chooses:

1. $f_1(x,y)=\sin^2x+\sin^2y$, with $f_1(0,0)=0$ and quadratic Taylor approximation $q_1(x,y)=x^2+y^2$.
2. $f_2(x,y)=(x+y)^2+x^4+y^4$, with $f_2(0,0)=0$ and quadratic Taylor approximation $q_2(x,y)=(x+y)^2$.

Both originals are nonnegative and have global minima at $(0,0)$. The second quadratic approximation has a flat direction along $x=-y$: its Hessian is positive semidefinite, not positive definite. The written source response contains an inconsistent first-derivative line for $\sin^2x+\sin^2y$; the runnable Python translation uses the functions and Taylor approximations shown in the original code cell.


```python
import numpy as np
import matplotlib.pyplot as plt

def f1(x,y): return np.sin(x)**2 + np.sin(y)**2
def q1(x,y): return x**2 + y**2
def f2(x,y): return (x+y)**2 + x**4 + y**4
def q2(x,y): return (x+y)**2

def compare_3d(original,quadratic,bounds,title):
    axis=np.linspace(*bounds,70)
    X,Y=np.meshgrid(axis,axis)
    fig=plt.figure(figsize=(9,6)); ax=fig.add_subplot(projection='3d')
    ax.plot_surface(X,Y,original(X,Y),color='C0',alpha=.65,linewidth=0)
    ax.plot_surface(X,Y,quadratic(X,Y),color='C1',alpha=.5,linewidth=0)
    from matplotlib.patches import Patch
    ax.legend(handles=[Patch(color='C0',label='Original'),Patch(color='C1',label='Quadratic')])
    ax.set(title=title,xlabel='x',ylabel='y',zlabel='z'); plt.show()

compare_3d(f1,q1,(-1.5,1.5),'Example 1')
compare_3d(f2,q2,(-1,1),'Example 2')
```

## Question 6 — Two functions with related quadratic approximations
The source defines
$$f(x,y)=2x^2+4xy+y^2,$$
$$g(x,y)=7+2(x+y)^2-y\sin y-x^3.$$
At the origin, $f$ equals its own quadratic Taylor approximation, and $g$ has Taylor polynomial
$$q_g(x,y)=7+2x^2+4xy+y^2=7+f(x,y).$$
Consequently both approximations have the same shape locally, differing by a vertical shift of 7. The original workbook plots the two original functions alongside their approximations in SageMath. Below is a runnable Python equivalent.


```python
def fq(x,y): return 2*x**2 + 4*x*y + y**2
def g(x,y): return 7 + 2*(x+y)**2 - y*np.sin(y) - x**3
def qg(x,y): return 7 + 2*x**2 + 4*x*y + y**2

compare_3d(fq,fq,(-1.5,1.5),'f is exactly quadratic')
compare_3d(g,qg,(-1.5,1.5),'g and its quadratic Taylor approximation')

# Near the expansion point, g-qg consists of cubic and higher-order terms:
assert np.isclose(g(0,0),qg(0,0))
```
