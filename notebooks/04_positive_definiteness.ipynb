## Question 3 — Positive eigenvalues and quadratic forms
For $A=\begin{bmatrix}a&b\\b&c\end{bmatrix}$, the submitted proof uses $a>0$ and $ac>b^2$: the product of eigenvalues is $\det A>0$ and their sum is $a+c>0$, so both eigenvalues are positive.

The four source matrices are
$$A_1=\begin{bmatrix}5&6\\6&7\end{bmatrix},\quad A_2=\begin{bmatrix}-1&-2\\-2&-5\end{bmatrix},$$
$$A_3=\begin{bmatrix}1&10\\10&100\end{bmatrix},\quad A_4=\begin{bmatrix}1&10\\10&101\end{bmatrix}.$$

**Submitted conclusions:** $A_1$ fails $ac>b^2$, $A_2$ has $a<0$, $A_3$ fails strict $ac>b^2$, and $A_4$ has positive eigenvalues. The source gives witnesses $x=(4,-3)$ for $A_1$ and $x=(1,0)$ for $A_2$, and observes the remaining two never have a strictly negative quadratic form. $A_3$ is positive **semidefinite**, not strictly positive definite.


```python
mats={
    'A1':np.array([[5.,6.],[6.,7.]]),
    'A2':np.array([[-1.,-2.],[-2.,-5.]]),
    'A3':np.array([[1.,10.],[10.,100.]]),
    'A4':np.array([[1.,10.],[10.,101.]])
}
for name,A in mats.items():
    eig=np.linalg.eigvalsh(A)
    label=('positive definite' if np.all(eig>1e-10) else
           'positive semidefinite' if np.all(eig>=-1e-10) else
           'negative definite' if np.all(eig<-1e-10) else 'indefinite')
    print(name,'det=',np.linalg.det(A),'eigenvalues=',eig,'classification=',label)
print('A1 at (4,-3):',np.array([4,-3])@mats['A1']@np.array([4,-3]))
print('A2 at (1,0):',np.array([1,0])@mats['A2']@np.array([1,0]))
```

## Question 4 — Are positive determinants sufficient?
**Submitted counterexample:** $A=-I=\begin{bmatrix}-1&0\\0&-1\end{bmatrix}$ has determinant $+1$ but is negative definite, corresponding to $-x^2-y^2$. Therefore positive determinant alone does not imply positive definiteness. For a **symmetric $2\times2$ matrix**, Sylvester's criterion requires $a>0$ and $ac-b^2>0$.


```python
A=-np.eye(2)
print('det(-I):',np.linalg.det(A),'eigenvalues:',np.linalg.eigvalsh(A))
assert np.linalg.det(A)>0 and np.all(np.linalg.eigvalsh(A)<0)
```

## Question 6 — Gradients of matrix polynomials
The source asks for gradients of $f(x)=b^Tx$ and $g(x)=x^TAx$, and whether the symmetry of $A$ changes the answer. The visible submitted answer gives $\nabla f(x)=b$; the screenshot cuts off immediately after the start of the $\nabla g$ response.

**Additional mathematical completion (not present in the visible source):**
$$\nabla_x(b^Tx)=b,\qquad \nabla_x(x^TAx)=(A+A^T)x.$$
For symmetric $A$, this simplifies to $2Ax$.


```python
def grad_linear(b):
    return np.asarray(b,dtype=float)

def grad_quadratic(A,x):
    A=np.asarray(A,dtype=float);x=np.asarray(x,dtype=float)
    return (A+A.T)@x

A=np.array([[2.,1.],[1.,2.]])
x=np.array([1.,2.]); b=np.array([3.,4.])
print('grad(bᵀx):',grad_linear(b))
print('grad(xᵀAx):',grad_quadratic(A,x),'= 2Ax:',2*A@x)
assert np.allclose(grad_quadratic(A,x),2*A@x)
```
