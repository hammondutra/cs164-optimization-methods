## Q1 — Understanding matrix inequalities

For a decision vector \(x\), \(Ax\leq b\) represents a collection of component-wise linear inequalities: each row of \(A\) gives coefficients and each element of \(b\) gives an upper bound. For instance, team effort limits \(x_1+x_2\leq100\), \(2x_1\leq80\) become

\[
\begin{bmatrix}1&1\\2&0\end{bmatrix}
\begin{bmatrix}x_1\\x_2\end{bmatrix}
\leq\begin{bmatrix}100\\80\end{bmatrix}.
\]

## Q2 — Convexity of linear programs

A linear program minimizes \(c^Tx\) subject to \(Ax\leq b\) and optionally \(Cx=d\). Affine functions satisfy
\(f(\lambda x+(1-\lambda)y)=\lambda f(x)+(1-\lambda)f(y)\): hence the objective is convex. Affine inequalities define convex half-spaces, affine equalities define convex hyperplanes, and their intersection is a convex feasible set.

## Q4 — Feasible but no finite maximizer

Maximize \(x+y\) subject to \(x,y\geq0\), \(-3x+2y\leq-1\), and \(x-y\leq2\).

The point (1,1) is feasible. Every \((t,t)\) with \(t\geq1\) is feasible as well, because \(-3t+2t=-t\leq-1\) and \(t-t=0\leq2\). But \(f(t,t)=2t\to\infty\). Therefore the LP is **feasible and unbounded**, with no finite optimum.

```r
region4(8)
region4(20)
```

## Q5 — Industrial optimization

Maximize \(8x+6y\) subject to \(x+y\leq2\), \(5x+10y\leq16\), \(2x+y\leq3\), \(x,y\geq0\).

| Vertex | Profit |
| --- | ---: |
| (0,0) | 0 |
| (1.5,0) | 12 |
| (1,1) | **14** |
| (0.8,1.2) | 13.6 |
| (0,1.6) | 9.6 |

The maximum profit is **14 at (1,1)**. The first and third constraints bind; the second has slack 1.

**Optional KKT verification:** Minimize \(-8x-6y\). The multipliers for the three resource constraints are \((4,0,2)\), and nonnegativity multipliers are zero. They satisfy stationarity and complementary slackness. Indeed, \(8x+6y=4(x+y)+2(2x+y)\leq14\), with equality at (1,1).

## Q6 — Speculative greedy algorithm

Start at a feasible vertex, inspect adjacent vertices, move to a neighbor with the greatest objective improvement, and repeat until no improving move is possible. This resembles **simplex**, though a complete method must handle feasibility search, degeneracy/cycling, and unbounded improving directions.

## Q8 — Norms

For \(x\in\mathbb R^n\):

- **L1:** \(\|x\|_1=\sum_i|x_i|\) — sum of absolute components.
- **L2:** \(\|x\|_2=\sqrt{\sum_i x_i^2}\) — Euclidean length.
- **L-infinity:** \(\|x\|_\infty=\max_i |x_i|\) — largest absolute component.
